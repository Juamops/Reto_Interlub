```python
import pandas as pd
import unicodedata
```

```python
data = pd.read_excel("../data/raw/DATOS_TEC.xlsx", sheet_name='Sheet1', index_col='Unnamed: 0')
data.head()
```

```python
data.info()
```

```python
cols = data.columns

def clean_text(text):
    text = '_'.join([elem for elem in text.lower().strip().split(' ') if '/' not in elem])
    text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode()
    
    return text
    
cols = [clean_text(c) for c in cols]
data.columns = cols
data = data[data.columns[:26]].reset_index()
aceites_cols = list(data.columns[2:14]) + list(data.columns[15:])
display(aceites_cols)
```
